```python
import wikipediaapi

# Wikipedia bağlantısını oluştur
wiki_wiki = wikipediaapi.Wikipedia('omadran@hacettepe.edu.tr')

# Aramak istediğiniz terimi girin
search_term = input("Wikipedia'da ne aramak istersiniz? ")

# Arama yap ve sonuçları göster
page = wiki_wiki.page(search_term)
if page.exists():
    print("Başlık:", page.title)
    print("URL:", page.fullurl)
    print("İçerik:")
    print(page.text[:500])  # Sadece ilk 500 karakteri göster
else:
    print("Aradığınız sayfa bulunamadı.")
```
